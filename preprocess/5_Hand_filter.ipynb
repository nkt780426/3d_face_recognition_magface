```python
output_file = 'final_session_remove.txt'

# 1. Đọc các đường dẫn đã có sẵn trong file (nếu file tồn tại)
existing_paths = set()
if os.path.exists(output_file):
    with open(output_file, 'r', encoding='utf-8') as f:
        for line in f:
            # Loại bỏ khoảng trắng và ký tự xuống dòng thừa
            existing_paths.add(line.strip())

# 2. Mở file ở chế độ append ('a') để ghi thêm những cái chưa có
added_count = 0
with open(output_file, 'a', encoding='utf-8') as f:
    for session_path, message in session_paths_to_add:
        # Kiểm tra xem session_path đã có trong file chưa (so sánh chuỗi đường dẫn)
        if session_path not in existing_paths:
            f.write(f"{session_path}\n")
            # Thêm vào set luôn để tránh ghi trùng nếu trong list truyền vào có phần tử bị lặp
            existing_paths.add(session_path) 
            added_count += 1

print(f"Đã kiểm tra xong! Thêm mới thành công {added_count} đường dẫn vào file '{output_file}'.")
```
